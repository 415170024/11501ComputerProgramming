### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 3：思考與分析（Questions）

1. 在函數內部，`scores` 的資料型態是什麼？其值為何？

2. 在函數內部，`options` 的資料型態是什麼？其值為何？
options 的資料型態是 dict（字典）。其值為

    'multiplier': 1.1,
    'bonus': 5
}
3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。
score * multiplier 是乘法，用來將分數乘以倍率；score ** 2 是次方運算，用來計算分數的平方。

4. 為什麼函數使用 `return`，而不是只使用 `print()`？
因為return 和 print() 的用途不同。
return：將計算結果交還給呼叫函數的地方，讓結果可以被儲存、運算或傳給其他函數;print()：將內容顯示在畫面上，但不會自動把顯示的內容當作函數的回傳值。
因此，本題使用 return 回傳分析結果字典，才能進一步透過 result["total"] 取得總分。

5. 如果呼叫 `analyze_scores(60, 70, 80)`，但沒有提供 `settings`，程式會如何執行？請說明原因。
程式仍然可以正常執行。因為函數中設定了預設值，如果沒有提供設定，multiplier 預設為 1，bonus 預設為 0，因此分數不會改變。

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

## 四、總結

在 Python 中，* 可以用於乘法、收集不定數量的位置參數，以及解包串列等可迭代物件；** 可以用於次方運算、收集不定數量的關鍵字參數，以及解包字典；return 則是將函數計算的結果回傳給呼叫端，方便後續使用。
例如：
 * 用於乘法
print(3 * 4)  # 結果：12

* ** 用於次方
print(2 ** 3)  # 結果：8

 * 用於收集位置參數
def show_scores(*scores):
    return scores

print(show_scores(60, 70, 80))  # 結果：(60, 70, 80)

* return 用於回傳結果
def square(number):
    return number ** 2

print(square(5))  # 結果：25
