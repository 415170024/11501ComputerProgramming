## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？因為 Python 的 int 使用任意精度，可以隨數字大小增加儲存空間。
2. Python 的 `int` 是否具有固定的位元數限制？沒有固定的位元數限制。
3. 什麼是 Arbitrary Precision（任意精度）？指整數可以根據需要使用更多記憶體，而不受固定位元數限制。
4. CPython 使用什麼物件來實作 `int`？PyLongObject。
5. Python 整數大小真正的限制是什麼？主要受到可用記憶體與系統資源限制。

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？因為 Python 的 float 使用二進位浮點數，0.1、0.2 無法精確表示，所以計算會產生誤差。
2. 浮點數是否能精確表示所有小數？不能。部分十進位小數只能近似表示。
3. 什麼是浮點數的精度（precision）？指浮點數能夠表示的有效數字數量。
4. 什麼是浮點數的範圍（range）？指浮點數能表示的最小值到最大值範圍。
5. 精度與範圍有何不同？精度是「準不準」，範圍是「能多大」。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？通常使用浮點數型別。
2. 什麼是 FP64？FP64 是 64 位元浮點數，精度較高，但需要較多記憶體。
3. 什麼是 FP32？FP32 是 32 位元浮點數，是深度學習中常見的浮點格式。
4. 什麼是 FP16？FP16 是 16 位元浮點數，比 FP32 使用更少記憶體，運算通常也更快，但精度與範圍較低。
5. 什麼是 BF16？BF16 是 16 位元浮點數，具有 8 位元指數和 7 位元尾數，因此範圍接近 FP32，但精度比 FP16 低。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？因為浮點數更適合表示模型需要的精細數值，而且低精度格式可以節省大量記憶體。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？指 int 沒有固定的位元數限制，可以隨數字大小增加儲存空間。
2. 它是否代表可以儲存無限大的整數？否。 實際上仍受到記憶體和系統資源限制。
3. CPython 如何實現這項特性？使用 PyLongObject，將大整數拆成多個內部 digit 儲存，需要時增加 digit 的數量。

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？因為 int 可以使用更多記憶體來儲存更大的整數。
2. 為什麼 `float` 卻存在範圍限制？因為 float 使用固定大小的浮點數格式，位元數有限。
3. 為什麼 `float` 可能產生誤差？因為許多十進位小數無法用二進位浮點數精確表示，只能用近似值儲存。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
   Python int：重視正確性，可以使用更多空間表示更大的整數。
   FP16、BF16：重視效率，用較少位元降低記憶體和運算成本。
2. 各自的優點與缺點。
   int：精確，但數值越大需要越多記憶體，運算也可能變慢。
   FP16、BF16：節省記憶體、運算速度快，但精度較低，可能產生數值誤差。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？因為模型參數非常多，使用 FP16 或 BF16 可以大幅降低記憶體需求與運算成本，提高訓練和推論效率。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
   1.需要更多記憶體。2.運算成本更高。3.訓練與推論速度可能降低。4.需要更高的硬體資源。

請用自己的話解釋：

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text

```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量

- GPU 運算效率

等角度進行討論。

Python 可以正常運作，是因為 int 使用任意精度，數字變大時可以使用更多記憶體來儲存，因此不受固定 32 位元或 64 位元的限制。
大型 AI 模型則不同，因為模型包含大量參數。如果每個參數都使用任意精度或高精度格式，就會需要非常大的記憶體，而且計算速度也會變慢。
AI 模型的權重通常不需要像 Python int 一樣追求完全精確，而是可以接受一定程度的數值誤差。因此使用 FP16、BF16 等較低精度格式，可以節省記憶體、提高運算速度，並充分利用 GPU 的低精度運算能力。
