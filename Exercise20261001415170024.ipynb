# 1. 0.1的二進位表示法
把十進位的小數轉成二進位，可以一直把小數部分乘以 2，然後取每次結果的整數部分。
## 計算過程
| 次數 | 計算 | 取出數字 | 小數部分 |
|---|---|---|---:|
| 1 | 0.1 × 2 = 0.2 | 0 | 0.2 |
| 2 | 0.2 × 2 = 0.4 | 0 | 0.4 |
| 3 | 0.4 × 2 = 0.8 | 0 | 0.8 |
| 4 | 0.8 × 2 = 1.6 | 1 | 0.6 |
| 5 | 0.6 × 2 = 1.2 | 1 | 0.2 |
| 6 | 0.2 × 2 = 0.4 | 0 | 0.4 |
| 7 | 0.4 × 2 = 0.8 | 0 | 0.8 |
| 8 | 0.8 × 2 = 1.6 | 1 | 0.6 |
| 9 | 0.6 × 2 = 1.2 | 1 | 0.2 |
| 10 | 0.2 × 2 = 0.4 | 0 | 0.4 |

把每一次得到的整數部分依序排列：

0.000110011001100110011...

因此：

**0.1 的二進位表示為：**

0.000110011001100110011...

## 看到的現象
從計算過程可以發現，0011 會一直重複出現：

0.000110011001100110011...

所以可以知道**0.1 的二進位表示是一個無限循環小數。**

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
Python 無法儲存無限多位的二進位數，因此會使用 IEEE 754 浮點數格式來表示。而 float 通常使用 64 位元的雙精度浮點數，但會將無法精確表示的數值取最接近的近似值。
IEEE 754 使用「四捨六入五成雙」的方式進行捨入，也就是取最接近的數值，如果剛好在兩個數值中間，就選擇最後一位是偶數的數值。

3. Python如何知道浮點數的精度？請問執行以下程式碼:
```python
import sys
print(sys.float_info)
```
並說明輸出結果的意思。
Python 可以透過 sys.float_info 查看目前系統中 float 浮點數的相關資訊。
各項意思如下：

- max：float 可以表示的最大值。

- min：float 可以表示的最小正數。

- dig=15：float 大約可以保證 15 位十進位有效數字。

- mant_dig=53：有效數字部分有 53 個二進位位元。

- epsilon：1.0 與下一個可表示浮點數之間的差距。

- radix=2：表示浮點數使用 二進位。

- rounds=1：表示使用「四捨五入到最接近值」的方式。

所以，從 sys.float_info 可以知道 Python 的 float 使用 IEEE 754 雙精度格式，而且有效精度大約是 15～17 位十進位數字。


４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```
可以使用 Python 的 round() 函數來處理:
`round(0.1 + 0.2, 1)`
結果會得到 0.3。不過這只是將結果四捨五入到指定的小數位數，並沒有改變浮點數本身的儲存方式。
